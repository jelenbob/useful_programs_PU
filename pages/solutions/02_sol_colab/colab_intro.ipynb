# Sumowanie ciągu geometrycznego

Ciąg geometryczny to ciąg liczb, w którym każdy kolejny wyraz powstaje przez pomnożenie poprzedniego przez stały iloraz $ q $.

Sumę pierwszych $ n $ wyrazów ciągu geometrycznego dla $$ q \neq 1 $$ obliczamy ze wzoru:

$$
S_n = a_1 \cdot \frac{1-q^n}{1-q}
$$

gdzie:
- $ a_1 $ – pierwszy wyraz ciągu,
- $ q $ – iloraz ciągu,
- $ n $ – liczba wyrazów.

### Przykład

Dla ciągu geometrycznego:

$$
2,\ 6,\ 18,\ 54
$$

mamy:

- $ a_1 = 2 $
- $ q = 3 $
- $ n = 4 $

Sumę można obliczyć bezpośrednio:

$$
S_4 = 2 + 6 + 18 + 54 = 80
$$

lub korzystając ze wzoru:

$$
S_4 = 2 \cdot \frac{1-3^4}{1-3}
= 2 \cdot \frac{1-81}{-2}
= 80
$$

Otrzymaliśmy sumę pierwszych czterech wyrazów równą 80.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, FFMpegWriter
from IPython.display import Video

# Parametry wahadła
L = 1.0
g = 9.81
theta0 = np.pi / 4  # 45 stopni

t = np.linspace(0, 10, 300)
omega = np.sqrt(g / L)

theta = theta0 * np.cos(omega * t)

x = L * np.sin(theta)
y = -L * np.cos(theta)

fig, ax = plt.subplots(figsize=(5, 5))
ax.set_xlim(-1.2, 1.2)
ax.set_ylim(-1.2, 0.2)
ax.set_aspect('equal')
ax.grid()

line, = ax.plot([], [], 'o-', lw=3)

def update(frame):
    line.set_data([0, x[frame]], [0, y[frame]])
    return line,

anim = FuncAnimation(fig, update, frames=len(t), interval=30)

video_path = "video.mp4"
writer = FFMpegWriter(fps=30)
anim.save(video_path, writer=writer)

plt.close(fig)

print(f"Zapisano: {video_path}")

# Wyświetlenie filmu w notebooku
Video(video_path, embed=True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Parametric equations
t = np.linspace(0, 2 * np.pi, 500)
x = np.exp(np.sin(t))
y = np.tan(t)
z = np.sqrt(t + 2)

# 3D Plot
fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, label="Parametric Curve")
ax.set_title("3D Parametric Curve", fontsize=14)
ax.set_xlabel("x(t)")
ax.set_ylabel("y(t)")
ax.set_zlabel("z(t)")
ax.legend()

# Zapis do pliku
plt.savefig("parametric_curve.png", dpi=300, bbox_inches="tight")

plt.show()

print("Plik zapisany jako: parametric_curve.png")
```
